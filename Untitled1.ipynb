## Model definition

### Subtask:
Define the Graph Neural Network model architecture, choosing appropriate layers (e.g., GCN, GAT, GraphSAGE).


**Reasoning**:
Import the necessary modules for defining the GNN model and define the GNN model class with two GCNConv layers, ReLU activation, and dropout. Instantiate the model.



```python
import torch
import torch.nn.functional as F
from torch_geometric.nn import GCNConv

class GCN(torch.nn.Module):
    def __init__(self, in_channels, hidden_channels, out_channels):
        super().__init__()
        self.conv1 = GCNConv(in_channels, hidden_channels)
        self.relu = torch.nn.ReLU()
        self.dropout = torch.nn.Dropout(p=0.5)
        self.conv2 = GCNConv(hidden_channels, out_channels)

    def forward(self, x, edge_index):
        x = self.conv1(x, edge_index)
        x = self.relu(x)
        x = self.dropout(x)
        x = self.conv2(x, edge_index)
        return x

# Instantiate the model
model = GCN(dataset.num_features, 16, dataset.num_classes)

print(model)
```

## Training

### Subtask:
Train the GNN model on the training data, defining a loss function and optimizer.


**Reasoning**:
Define the loss function and optimizer, and then implement the training loop for the specified number of epochs.



```python
import torch.optim as optim

# 1. Define the loss function
criterion = torch.nn.CrossEntropyLoss()

# 2. Define the optimizer
optimizer = optim.Adam(model.parameters(), lr=0.01)

# 4. Train the model for a specified number of epochs
epochs = 200
for epoch in range(epochs):
    # 3a. Set the model to training mode
    model.train()

    # 3b. Zero the gradients of the optimizer
    optimizer.zero_grad()

    # 3c. Perform a forward pass
    out = model(data.x, data.edge_index)

    # 3d. Calculate the loss
    loss = criterion(out[data.train_mask], data.y[data.train_mask])

    # 3e. Perform a backward pass
    loss.backward()

    # 3f. Update the model's weights
    optimizer.step()

    # Print the loss every few epochs
    if (epoch + 1) % 10 == 0:
        print(f'Epoch: {epoch+1:03d}, Loss: {loss:.4f}')

print("Training finished.")
```

## Evaluation

### Subtask:
Evaluate the trained model on the validation and test sets to assess its performance.


**Reasoning**:
Evaluate the trained model on the validation and test sets by setting the model to evaluation mode, performing a forward pass, getting predictions, calculating accuracies for training, validation, and test sets using the respective masks, and finally printing the accuracies.



```python
# 1. Set the model to evaluation mode
model.eval()

# 2. Perform a forward pass with the entire graph data
out = model(data.x, data.edge_index)

# 3. Get the predicted class for each node
pred = out.argmax(dim=1)

# 4. Calculate the accuracy on the training, validation, and test sets
train_correct = pred[data.train_mask] == data.y[data.train_mask]
train_acc = int(train_correct.sum()) / int(data.train_mask.sum())

val_correct = pred[data.val_mask] == data.y[data.val_mask]
val_acc = int(val_correct.sum()) / int(data.val_mask.sum())

test_correct = pred[data.test_mask] == data.y[data.test_mask]
test_acc = int(test_correct.sum()) / int(data.test_mask.sum())

# 5. Print the training, validation, and test accuracies
print(f'Train Accuracy: {train_acc:.4f}')
print(f'Validation Accuracy: {val_acc:.4f}')
print(f'Test Accuracy: {test_acc:.4f}')
```

## Prediction

### Subtask:
Use the trained model to make predictions on new, unseen graph data.


**Reasoning**:
Use the trained model to make predictions on the entire graph data following the instructions.



```python
import torch.nn.functional as F

# Ensure the trained model is in evaluation mode
model.eval()

# Perform a forward pass with the trained model on the entire graph data
logits = model(data.x, data.edge_index)

# Apply the softmax function to the output logits to obtain class probabilities
probabilities = F.softmax(logits, dim=1)

# Get the predicted class label for each node
predictions = torch.argmax(probabilities, dim=1)

# The resulting tensor 'predictions' contains the predicted class labels for all nodes.
# You can inspect the predictions, for example, by printing the first few.
print("Sample predictions:", predictions[:10])
```

## Visualization

### Subtask:
Visualize the graph and the model's predictions to gain insights.


**Reasoning**:
Import necessary libraries for visualization and convert the PyTorch Geometric graph to a NetworkX graph, assigning true and predicted labels as node attributes.



```python
import networkx as nx
import matplotlib.pyplot as plt

# Convert PyTorch Geometric graph to NetworkX graph
# PyG edge_index is a 2xN tensor, need to convert to list of tuples for NetworkX
edge_list = data.edge_index.t().tolist()
G = nx.Graph(edge_list)

# Ensure all nodes from data.x are in the graph, even if they have no edges
# NetworkX might exclude nodes without edges if only initialized from edge_list
if G.number_of_nodes() < data.num_nodes:
    missing_nodes = set(range(data.num_nodes)) - set(G.nodes())
    G.add_nodes_from(missing_nodes)


# Assign true labels as node attributes
true_labels = data.y.tolist()
for i, label in enumerate(true_labels):
    G.nodes[i]['true_label'] = label

# Assign predicted labels as node attributes
predicted_labels = predictions.tolist()
for i, label in enumerate(predicted_labels):
    G.nodes[i]['predicted_label'] = label

print("NetworkX graph created and attributes assigned.")
print(f"Number of nodes in NetworkX graph: {G.number_of_nodes()}")
print(f"Number of edges in NetworkX graph: {G.number_of_edges()}")
```

**Reasoning**:
Visualize the NetworkX graph by coloring nodes based on the true labels, using a layout algorithm for placement and adding a title.



```python
# Create a layout for visualizing the graph
pos = nx.spring_layout(G, seed=42) # Use a fixed seed for reproducibility

# Map integer labels to colors
# There are 7 classes in the Cora dataset (0-6)
label_to_color = {
    0: 'red',
    1: 'blue',
    2: 'green',
    3: 'purple',
    4: 'orange',
    5: 'brown',
    6: 'pink'
}

# Get node colors based on true labels
node_colors_true = [label_to_color[G.nodes[i]['true_label']] for i in G.nodes()]

# Draw the graph colored by true labels
plt.figure(figsize=(12, 12))
nx.draw(G, pos, node_color=node_colors_true, node_size=20, with_labels=False, width=0.5)
plt.title("Cora Graph with True Labels", fontsize=16)
plt.show()
```

**Reasoning**:
Visualize the NetworkX graph by coloring nodes based on the predicted labels, using the same layout as before, and adding a title for comparison with the true labels visualization.



```python
# Get node colors based on predicted labels
node_colors_predicted = [label_to_color[G.nodes[i]['predicted_label']] for i in G.nodes()]

# Draw the graph colored by predicted labels using the same layout
plt.figure(figsize=(12, 12))
nx.draw(G, pos, node_color=node_colors_predicted, node_size=20, with_labels=False, width=0.5)
plt.title("Cora Graph with Predicted Labels", fontsize=16)
plt.show()
```
